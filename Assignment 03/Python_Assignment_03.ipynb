#Problem Statement
To check whether or not the triangle is right angled triangle using the function
#Aim
Write a python program that accepts a length of three sides of triangle as inputs. The program should indicate whether or not the triangle is a right angled triangle using function

```python
def is_right_angled(a, b, c):
    sides = sorted([a, b, c])
    return sides[0]**2 + sides[1]**2 == sides[2]**2

a = float(input("Enter first side: "))
b = float(input("Enter second side: "))
c = float(input("Enter third side: "))

if is_right_angled(a, b, c):
    print("The triangle is right-angled.")
else:
    print("The triangle is not right-angled.")
```
