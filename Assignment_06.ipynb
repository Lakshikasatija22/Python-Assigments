```python
import pandas as pd

# Create a sample DataFrame
data = {
    'Name': ['Amit', 'Priya', 'Rahul', 'Sneha', 'Amit'],
    'Marks': [85, 90, None, 78, 85],
    'City': ['Pune', 'Mumbai', 'Delhi', 'Pune', 'Pune'],
    'Age': [19, 18, 19, 18, 19]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Display first five rows
print("\nFirst five rows:")
print(df.head())

# Display information about the dataset
print("\nDataset Information:")
df.info()

# Display statistical summary
print("\nStatistical Summary:")
print(df.describe())

# Identify missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Fill missing marks with the mean
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())

# Identify duplicate records
print("\nDuplicate Records:")
print(df.duplicated())

# Remove duplicate records
df = df.drop_duplicates()

# Remove the irrelevant Age column
df = df.drop(columns=['Age'])

# Select students who scored more than 80 marks
print("\nStudents scoring more than 80:")
print(df.loc[df['Marks'] > 80])

# Display the final cleaned dataset
print("\nFinal Cleaned DataFrame:")
print(df)
```
