```python
#List
L=[30,10,20,40]
print("List:",L)
L.append(50)
L.insert(1,15)
L.remove(20)
L.pop()
L.sort()
print("List after operations :",L)

#Tuple
T=(10,20,10,30)
print("\nTuple:",T)
print("Count:",T.count(10))
print("Index:",T.index(20))
print("First Element:",T[0])
print("Slice:",T[1:3])
print("Concatenation:",T+(40,))

#Dictionary
D={"name":"Kanak","age":18,"branch":"CSE"}
print("\nDictionary:",D)
D["city"]="Pune"
D.update({"age":19})
print("keys:",D.keys())
print("Values:",D.values())
print("Name:",D.get("name"))
```
